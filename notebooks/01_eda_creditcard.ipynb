### Para verificação de possíveis outliers

```python
#  Para o Amount

plt.boxplot(df["Amount"])
plt.title("Boxplot do Valor das Transações")
plt.show()
```

```python
# Para o Time

plt.boxplot(df["Time"])
plt.title("Boxplot do Tempo")
plt.show()
```

```python
# ======================================================
# Para a quantidade de transações normais e fraudulentas 
# ======================================================

df["Class"].value_counts()
```

### Para ver a proporção de Fraude x Normal

```python
labels = ["Normal", "Fraude"]
sizes = df["Class"].value_counts()

plt.pie(sizes, labels=labels, autopct="%1.2f%%", colors=["#66b3ff", "#ff6666"])
plt.title("Proporção de Transações")
plt.show()
```
